```python
# Importing various libraries for data handling and deep learning  

import os
import time
import copy
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

from torchvision import transforms
from torchvision.models import (
    resnet50, ResNet50_Weights
)
```

```python
# Setting the dataset path  

random.seed(42)

dataset_dir = r"D:\College\Masters\CS7050_Artificial Intelligence\food-101\food-101"

image_dir = os.path.join(dataset_dir, "images")
meta_dir = os.path.join(dataset_dir, "meta")

print("Image folder exists:", os.path.exists(image_dir))
print("Meta folder exists:", os.path.exists(meta_dir))
```

```python
# Loading metadata files to obtain class names as well as training and testing image paths

classes_file = os.path.join(meta_dir, "classes.txt")
train_file = os.path.join(meta_dir, "train.txt")
test_file = os.path.join(meta_dir, "test.txt")

with open(classes_file, "r") as f:
    class_names = [line.strip() for line in f.readlines()]

with open(train_file, "r") as f:
    train_paths = [line.strip() for line in f.readlines()]

with open(test_file, "r") as f:
    test_paths = [line.strip() for line in f.readlines()]

print("Number of classes:", len(class_names))
print("Number of train entries:", len(train_paths))
print("Number of test entries:", len(test_paths))
print("First 5 classes:", class_names[:5])
```

```python
# Mapping class names to numerical labels

class_to_idx = {class_name: idx for idx, class_name in enumerate(class_names)}
idx_to_class = {idx: class_name for class_name, idx in class_to_idx.items()}


# Sample Mapping to verify

print("Example class mapping:")
for i, (k, v) in enumerate(class_to_idx.items()):
    print(f"{k} -> {v}")
    if i == 4:
        break
```

```python
# Creating dataframes to store image paths and labels with the help of a function

def build_dataframe(path_list, image_dir, class_to_idx):
    records = []

    for rel_path in path_list:
        class_name = rel_path.split("/")[0]
        image_path = os.path.join(image_dir, rel_path + ".jpg")
        label = class_to_idx[class_name]

        records.append({
            "relative_path": rel_path,
            "image_path": image_path,
            "class_name": class_name,
            "label": label
        })

    return pd.DataFrame(records)


train_df_full = build_dataframe(train_paths, image_dir, class_to_idx)
test_df = build_dataframe(test_paths, image_dir, class_to_idx)

print("Train dataframe shape:", train_df_full.shape)
print("Test dataframe shape:", test_df.shape)
```

```python
train_df_full.head()
```

```python
# Splitting training data into training and validation datasets

train_df, val_df = train_test_split(
    train_df_full,
    test_size=0.2,
    random_state=42,
    stratify=train_df_full["label"]
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("Train split:", train_df.shape)
print("Validation split:", val_df.shape)
print("Test split:", test_df.shape)
```

```python
## Defining image preprocessing and augmentation transformers 

IMG_SIZE = 224
BATCH_SIZE = 16

train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
# Creating a custom dataset class to load images and labels

class Food101CustomDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, idx):
        row = self.dataframe.iloc[idx]

        image = Image.open(row["image_path"]).convert("RGB")
        label = int(row["label"])

        if self.transform:
            image = self.transform(image)

        return image, label
```

```python
# Creating datasets and dataloaders for training, validation and testing

train_dataset = Food101CustomDataset(train_df, transform=train_transform)
val_dataset = Food101CustomDataset(val_df, transform=eval_transform)
test_dataset = Food101CustomDataset(test_df, transform=eval_transform)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
# Function for training the model for one epoch and returning loss and accuracy
# Training only one epoch for Quick performance check
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()

    running_loss = 0.0
    all_preds = []
    all_labels = []

    for images, labels in loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)

        preds = torch.argmax(outputs, dim=1)
        all_preds.extend(preds.detach().cpu().numpy())
        all_labels.extend(labels.detach().cpu().numpy())

    epoch_loss = running_loss / len(loader.dataset)
    epoch_acc = accuracy_score(all_labels, all_preds)

    return epoch_loss, epoch_acc


# Fuction for evaluating the model on validation and test data without updating weights

def evaluate(model, loader, criterion, device):
    model.eval()

    running_loss = 0.0
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)

            preds = torch.argmax(outputs, dim=1)
            all_preds.extend(preds.detach().cpu().numpy())
            all_labels.extend(labels.detach().cpu().numpy())

    epoch_loss = running_loss / len(loader.dataset)
    epoch_acc = accuracy_score(all_labels, all_preds)

    return epoch_loss, epoch_acc



# Function for running the full training loop across multiple epochs and tracking the performance

def train_model(model, train_loader, val_loader, criterion, optimizer, device, epochs=3):
    history = {
        "train_loss": [],
        "train_acc": [],
        "val_loss": [],
        "val_acc": []
    }

    best_model_wts = copy.deepcopy(model.state_dict())
    best_val_acc = 0.0

    start_time = time.time()

    for epoch in range(epochs):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc = evaluate(model, val_loader, criterion, device)

        history["train_loss"].append(train_loss)
        history["train_acc"].append(train_acc)
        history["val_loss"].append(val_loss)
        history["val_acc"].append(val_acc)

        print(
            f"Epoch [{epoch+1}/{epochs}] | "
            f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.4f} | "
            f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.4f}"
        )

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_model_wts = copy.deepcopy(model.state_dict())

    total_time = time.time() - start_time

    model.load_state_dict(best_model_wts)

    return model, history, best_val_acc, total_time
```

```python
# Function for plotting training and validation loss and accuracy over epochs

def plot_history(history, title):
    epochs = range(1, len(history["train_loss"]) + 1)

    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.plot(epochs, history["train_loss"], label="Train Loss")
    plt.plot(epochs, history["val_loss"], label="Validation Loss")
    plt.title(f"{title} Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(epochs, history["train_acc"], label="Train Accuracy")
    plt.plot(epochs, history["val_acc"], label="Validation Accuracy")
    plt.title(f"{title} Accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()

    plt.tight_layout()
    plt.show()
```

```python
# Defining key training hyperparameters

NUM_CLASSES = 101
LEARNING_RATE = 0.0001
EPOCHS = 3
```

```python
# Function for freezing pretrained model layers to prevent weight updates

def freeze_model(model):
    for param in model.parameters():
        param.requires_grad = False

        
        
# Function for creating a custom classification head  
        
def build_custom_head(in_features, num_classes=101):
    return nn.Sequential(
        nn.Linear(in_features, 512),
        nn.ReLU(),
        nn.Dropout(0.3),

        nn.Linear(512, 256),
        nn.ReLU(),
        nn.Dropout(0.3),

        nn.Linear(256, num_classes)
    )


# Loading pretrained ResNet model and replacing its classifier with a custom head

def get_resnet50_model(num_classes=101):
    model = resnet50(weights=ResNet50_Weights.DEFAULT)

    freeze_model(model)

    in_features = model.fc.in_features
    model.fc = build_custom_head(in_features, num_classes)

    return model
```

```python
# Unfreezing layer3, layer4, and classifier head for fine-tuning

def unfreeze_layer3_layer4(model):
    for param in model.layer3.parameters():
        param.requires_grad = True

    for param in model.layer4.parameters():
        param.requires_grad = True

    for param in model.fc.parameters():
        param.requires_grad = True

    return model
```

```python
# Loading pretrained ResNet model and replacing its classifier with a custom head

def get_resnet_model(num_classes=101):
    model = resnet50(weights=ResNet50_Weights.DEFAULT)
    freeze_model(model)

    in_features = model.fc.in_features
    model.fc = build_custom_head(in_features, num_classes)

    return model
```

```python
#  Training and evaluating the ResNet model on validation and test data(1st)

resnet_model = get_resnet_model(NUM_CLASSES).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(
    filter(lambda p: p.requires_grad, resnet_model.parameters()),
    lr=LEARNING_RATE
)

resnet_model, resnet_history, resnet_best_val_acc, resnet_train_time = train_model(
    resnet_model,
    train_loader,
    val_loader,
    criterion,
    optimizer,
    device,
    epochs=EPOCHS
)

resnet_test_loss, resnet_test_acc = evaluate(resnet_model, test_loader, criterion, device)

print("\nResNet Test Loss:", round(resnet_test_loss, 4))
print("ResNet Test Accuracy:", round(resnet_test_acc, 4))
print("ResNet Training Time (seconds):", round(resnet_train_time, 2))
```

```python
# Plotting training and validation loss and accuracy for ResNet model

plot_history(resnet_history, "ResNet50")
```

```python
# Finetuning ResNet with layer3, layer4, and classifier head unfrozen

resnet_layer3_4_model = get_resnet50_model(NUM_CLASSES).to(device)
resnet_layer3_4_model = unfreeze_layer3_layer4(resnet_layer3_4_model)

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    filter(lambda p: p.requires_grad, resnet_layer3_4_model.parameters()),
    lr=LEARNING_RATE
)

resnet_layer3_4_model, layer3_4_history, layer3_4_best_val_acc, layer3_4_train_time = train_model(
    resnet_layer3_4_model,
    train_loader,
    val_loader,
    criterion,
    optimizer,
    device,
    epochs=EPOCHS
)

layer3_4_test_loss, layer3_4_test_acc = evaluate(
    resnet_layer3_4_model,
    test_loader,
    criterion,
    device
)

print("\nStrategy 2 - Layer3+Layer4 Test Loss:", round(layer3_4_test_loss, 4))
print("Strategy 2 - Layer3+Layer4 Test Accuracy:", round(layer3_4_test_acc, 4))
print("Strategy 2 - Training Time (seconds):", round(layer3_4_train_time, 2))
```

```python
# Plotting training and validation performance for the second strategy

plot_history(layer3_4_history, "ResNet50 Fine-Tuning - Layer3 and Layer4")
```
